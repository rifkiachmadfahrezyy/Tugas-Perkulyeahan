**MENGGANTI NAMA PADA VARIABEL AGAR LEBIH MENDESKRIPSIKAN APA YANG ADA DI DALAMNYA**

```python
df_gt = (
    df_tpt[['Provinsi', 'tpt_avg']]
    .merge(df_pop[['Provinsi', 'Population (Thousand)']], on='Provinsi')
    .merge(df_mskn[['Provinsi', 'Total Annually (Average)']], on='Provinsi')
)
df_gt['Total Annually (Average)'] = df_gt['Total Annually (Average)'].astype(float)
df_gt['Population (Thousand)'] = df_gt['Population (Thousand)'].astype(float)
df_gt = df_gt.rename(columns={"Total Annually (Average)" : "Annual poverty rate avg (Thousand)"})
df_gt = df_gt.rename(columns={"tpt_avg" : "Annually open unemployment rate avg (Percent)"})
df_gt
```

**MEMBUAT VARIABEL BARU PADA TABEL df_gt**

```python
df_gt['Annually Poverty rate avg (Percent)'] = (
    df_gt['Annual poverty rate avg (Thousand)']
    / df_gt['Population (Thousand)'] * 100
)

df_gt
```

```python
print(df_gt[['Annually open unemployment rate avg (Percent)',
                         'Population (Thousand)',
                         'Annually Poverty rate avg (Percent)']].describe())
```

```python
A = df_gt['Annually open unemployment rate avg (Percent)'] > 5
B = df_gt['Annually Poverty rate avg (Percent)'] > 10
semesta = len(df_gt)
print(f'A = {A.sum()} \nB = {B.sum()}')
```

```python
tot_prov = semesta
Peluang_A = A.sum() / tot_prov
Peluang_B = B.sum() / tot_prov
print(f"P(A) bernilai : {Peluang_A:.3f}")
print(f"P(B) bernilai : {Peluang_B:.3f}")
```

```python
both_true = (A & B)
one_true = (A | B)
print(f"jumlah provinsi yang termasuk dalam kedua himpunan ada sebanyak : {both_true.sum()}")
print(f"jumlah provinsi yang termasuk dalam salah satu himpunan atau lebih ada sebanyak : {one_true.sum()}")
Peluang_A_irisan_B = both_true.sum() / tot_prov 
print(f"P(A ∩ B) bernilai : {Peluang_A_irisan_B:.3f}")
```

```python
Peluang_bersyarat_A_dan_B = both_true.sum() / B.sum()
print(f"Hasil dari P(A | B) adalah :{Peluang_bersyarat_A_dan_B}")
```

```python
pel_A_x_pel_B = Peluang_A * Peluang_B
print(f"P(A ∩ B) : {Peluang_A_irisan_B}")
print(f"P(A) × P(B) : {pel_A_x_pel_B}")
selisih = (pel_A_x_pel_B - Peluang_A_irisan_B) * 38
print(selisih)
```

```python
from scipy.stats import hypergeom
import pandas as pd

N = 38 # jumlah provinsi di indonesia
K = 16 # Jumlah provinsi di indonesia yang memiliki tingkat kemiskinan diatas 10%
n = 6

x = range(0, 7)

p = hypergeom.pmf(x, N, K, n)

hasil = pd.DataFrame({
    'Jumlah provinsi miskin': x,
    'Probabilitas': p
})

print(hasil.round(7).to_string(index=False))
```

```python
p_tepat3 = hypergeom.pmf(2, N, K, n)
p_maks3 = hypergeom.cdf(4, N, K, n)
print(p_tepat3)
print(p_maks3)
p_maks3 = p_maks3 * 100
p_tepat3 = p_tepat3 * 100
print(f"{p_tepat3:.2f}")
print(f"{p_maks3:.2f}")
```

```python
N = 38
K = 16
n = 6

x = np.arange(0, n + 1)

p = hypergeom.pmf(x, N, K, n)

print("Nilai X :", x)
print("P(X=x)  :", p)
print("Jumlah probabilitas :", p.sum())

plt.bar(x, p, edgecolor='black')

plt.xlabel("Jumlah Provinsi dengan Kemiskinan di atas 10%")
plt.ylabel("P(X = x)")
plt.title("Distribusi Hipergeometrik")

plt.show()
```

```python
jumlah_simulasi = 1000

hasil_simulasi = hypergeom.rvs(
    M=N,
    n=K,
    N=n,
    size=jumlah_simulasi
)

mean_simulasi = np.mean(hasil_simulasi)
mean_teoritis = n * (K / N)

print("Mean simulasi :", mean_simulasi)
print("Mean teoritis :", mean_teoritis)
```
